```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Load Dataset
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date']) 
```

```python
df_ARG= df[df['job_country'] == 'Argentina'].copy()

df_ARG
```

```python
df_ARG['job_posted_month'] = df_ARG['job_posted_date'].dt.strftime('%B')

df_ARG
```

```python
df_ARG_pivot = df_ARG.pivot_table(index='job_posted_month', columns='job_title_short', aggfunc= 'size')
```

```python
df_ARG_pivot.reset_index (drop=True, inplace= True)
df_ARG_pivot['month_no'] = pd.to_datetime(df_ARG_pivot['job_posted_month'], format= '%B').dt.month
df_ARG_pivot.sort_values('month_no', inplace= True)
df_ARG_pivot.set_index('job_posted_month', inplace= True)
df_ARG_pivot.drop(columns= 'month_no', inplace= True)
```

```python
df_ARG_pivot.drop(columns=['level_0', 'index'], inplace=True)
```

```python
df_ARG_pivot
```

```python
df_ARG_pivot.plot(kind='line')
```

```python
top_3= df_ARG['job_title_short'].value_counts().head(3)
top_3= top_3.index.tolist()
```

```python
df_ARG_pivot[top_3].plot(kind='line')
plt.xlabel('2023')
plt.ylabel('Job Count')
plt.legend()
plt.show()
```
